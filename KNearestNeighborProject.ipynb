```python
import pandas as pd

ratings = pd.read_csv(
    "u.data", sep="\t", names=["userId", "movieId", "rating", "timestamp"]
)

ratings
```

```python
movies = pd.read_csv(
    "u.item",
    sep="|",
    encoding="latin-1",
    names=[
        "movieId",
        "title",
        "releaseDate",
        "videoReleaseDate",
        "IMDbURL",
        "genreUnknow",
        "genreAction",
        "genreAdventure",
        "genreAnimation",
        "genreChildren",
        "genreComedy",
        "genreCrime",
        "genreDocumentary",
        "genreDrama",
        "genreFantasy",
        "genreFilmNoir",
        "genreHorror",
        "genreMusical",
        "genreMystery",
        "genreRomance",
        "genreSciFi",
        "genreThriller",
        "genreWar",
        "genreWestern",
    ],
)

movies
```

```python
data = ratings.merge(movies[["movieId", "title"]], on="movieId", how="left")
data
```

```python
data["title"].isna().sum()
```

```python
client_data = pd.pivot_table(
    data=data, values="rating", index="userId", columns="title", fill_value=0
)
client_data
```

On va faire notre modéle de machine learning à partir de ce jeu de donnée. Pour faire nos recommendation de film on va essayer de trouver les n plus proche voisins pour chaque client.

```python
from sklearn.neighbors import NearestNeighbors

knn = NearestNeighbors(metric="cosine")
knn.fit(client_data)
```

La premiére chose qu'on a besoin est de créer des listes de film non vue par les utilisateurs.

```python
def film_non_vue(data, client_data, userId):
    films = set(client_data.columns)
    print(films)
    client_film = set(data.query(f"`userId` == {userId}")["title"])
    print(client_film)
    return films - client_film


film_non_vue(data, client_data, 1)
```

```python
distances, indices = knn.kneighbors(
    client_data.loc[1].to_numpy().reshape(1, -1), n_neighbors=6
)
distances
```
