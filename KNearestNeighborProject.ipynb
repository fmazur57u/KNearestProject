```python
import pandas as pd

ratings = pd.read_csv(
    "u.data", sep="\t", names=["userId", "movieId", "rating", "timestamp"]
)

ratings
```

```python
movies = pd.read_csv(
    "u.item",
    sep="|",
    encoding="latin-1",
    names=[
        "movieId",
        "title",
        "releaseDate",
        "videoReleaseDate",
        "IMDbURL",
        "genreUnknow",
        "genreAction",
        "genreAdventure",
        "genreAnimation",
        "genreChildren",
        "genreComedy",
        "genreCrime",
        "genreDocumentary",
        "genreDrama",
        "genreFantasy",
        "genreFilmNoir",
        "genreHorror",
        "genreMusical",
        "genreMystery",
        "genreRomance",
        "genreSciFi",
        "genreThriller",
        "genreWar",
        "genreWestern",
    ],
)

movies
```

```python
data = ratings.merge(movies[["movieId", "title"]], on="movieId", how="left")
data
```

```python
data["title"].isna().sum()
```

```python
client_data = pd.pivot_table(
    data=data, values="rating", index="userId", columns="title", fill_value=0
)
client_data
```

On va faire notre modéle de machone learning à partir de ce jeu de donnée. Pour analyser ce je ude donnée on va d'abords faire une réduction de dimension. Comme toutes les colonnes sont de même échelle on peut utiliser une analyse en composante principale directement sans standardisation.

```python
from sklearn.decomposition import PCA
from sklearn import set_config

set_config(transform_output="pandas")

pca = PCA(n_components=3)

red_data = pca.fit_transform(client_data)
red_data
```

```python
import seaborn as sns
import plotly.express as px

fig = px.scatter_3d(red_data, x="pca0", y="pca1", z="pca2", size_max=18)
fig.update_traces(marker_size=3)
fig.show()
```

On observe qu'il n'y a pas vraiment de groupes de client différents mais qu'il y a une zone dans lequel il y a beaucoup de clients qui sont proche puis plus on s'éloigne selon les trois composante principale, les clients sont de moins en moins proche. Essayons de voir si une autres méthodes de réduction de dimension peut nous montrer des cluster distinct. Essayons le tsne.

```python
from sklearn.manifold import TSNE

red_data = TSNE(n_components=3, perplexity=40).fit_transform(client_data)
red_data
```

On voit que le tsne ne révéle aucun cluster. Testons la méthode ISOMAP avec la métrique cosinus.

```python
fig = px.scatter_3d(red_data, x="tsne0", y="tsne1", z="tsne2")
fig.update_traces(marker_size=3)
fig.show()
```

```python
from sklearn.manifold import Isomap

red_data = Isomap(n_components=3, n_neighbors=3, metric="cosine").fit_transform(
    client_data
)
red_data
```

```python
fig = px.scatter_3d(red_data, x="isomap0", y="isomap1", z="isomap2")
fig.update_traces(marker_size=3)
fig.show()
```

La méthode Isomap nous permet de voir qu'il y a selon la métrique cosinus trois cluster de client.
